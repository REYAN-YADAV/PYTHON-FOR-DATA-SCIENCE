# Operators in Python

## Operators and Expression in Python

- An **operator** is a symbol which is used to perform some operations on objects or values.
- If any operator connects with two or more objects or values, then it is called an **expression**.
- In Python programming, we have **7 types of operators**. They are:

1. Arithmetic operators
2. Assignment operators
3. Relational operators (comparison)
4. Logical operators (comparison)
5. Bitwise operators
6. Membership operators
   - a) `in` operator
   - b) `not in` operator
7. Identity operator
   - a) `is` operator
   - b) `is not` operator

**Note 1:** Python programming does not support unary operators like `++` and `--`.

**Note 2:** Python does not support ternary operator (`res = Exp1 ? Exp2 : Exp3`).

**Note 3:** Python supports shorthand operators (`+=`, `-=`, `*=`, etc.).

**Note 4:** Python supports its own ternary operator called the **"if else operator"**.

## 1. Arithmetic Operators

- Arithmetic operators are those which are used to perform arithmetic operations such as Add, Sub, Mult, etc.
- If an arithmetic operator connects with two or more objects or values, then it is called an **arithmetic expression**.
- In Python programming we have **7 types** of arithmetic operators.

| SL No | Symbol | Meaning | Example (a=10, b=3) | Result |
|---|---|---|---|---|
| 1 | `+` | Addition | `print(a+b)` | 13 |
| 2 | `-` | Subtraction | `print(a-b)` | 7 |
| 3 | `*` | Multiplication | `print(a*b)` | 30 |
| 4 | `/` | Division (float quotient) | `print(a/b)` | 3.3333333333333335 |
| 5 | `//` | Floor division (integer quotient) | `print(a//b)` | 3 |
| 6 | `%` | Modulo division (remainder division) | `print(a%b)` | 1 |
| 7 | `**` | Exponentiation (power) | `print(a**b)` | 1000 |

````python
a, b = 10, 3
print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)
print("a // b =", a // b)
print("a % b  =", a % b)
print("a ** b =", a ** b)
````

**Note:**
```
print(10/3)   -> 3.3333333333333335
print(10//3)  -> 3
```

````python
print(10 / 3)
print(10 // 3)
````

## 2. Assignment Operator

- The purpose of the assignment operator is to **assign an RHS value / expression** to the LHS variable, i.e. to transfer the RHS value/expression value to the LHS variable.
- The symbol for the assignment operator is a single equal-to (`=`).
- In Python programming, we can use the assignment operator in two ways:

1. Single line assignment
2. Multi line assignment

### 1. Single line Assignment

**Syntax:**
```
LHS varname = RHS value
LHS varname = RHS Expression
```

With single line assignment, at a time we can assign one RHS value/expression to a single LHS variable name.

**Example:**
```
>>> a = 10
>>> b = 20
>>> c = a + b
>>> print(a, b, c) -> 10 20 30
```

````python
a = 10
b = 20
c = a + b
print(a, b, c)
````

### 2. Multi line Assignment

**Syntax:**
```
var1, var2, ... , var-n = Expr1, Expr2, ... , Expr-n
```

- Here the values of Val1, Val2, ..., Val-n are assigned to var1, var2, ..., var-n respectively.
- Here the values of Expr1, Expr2, ..., Expr-n are assigned to var1, var2, ..., var-n respectively.

**Examples:**
```
>>> a, b = 10, 20
>>> print(a, b)          -> 10 20
>>> c, d, e = a+b, a-b, a*b
>>> print(c, d, e)       -> 30 -10 200

>>> a, b = 10, 20
>>> a, b = b, a          # swapping logic
>>> print(a, b)          -> 20 10
```

````python
a, b = 10, 20
print(a, b)

c, d, e = a + b, a - b, a * b
print(c, d, e)

# swapping logic
a, b = 10, 20
a, b = b, a
print(a, b)
````

## 3. Relational Operators (Comparison Operators)

- The purpose of the relational operator is to compare two object values.
- If a relational operator connects with two object values, then it is called a **relational expression**.
- The result of a relational expression is either `True` or `False`.
- The relational expression is called a **test condition**.
- In Python programming there are **6 types** of relational operators. They are:

| S.No | Symbol | Meaning | Examples |
|---|---|---|---|
| 1 | `>` | greater than | `print(10>5) -> True`, `print(10>20) -> False` |
| 2 | `<` | less than | `print(10<5) -> True`, `print(10<20) -> False` |
| 3 | `==` | equality (double equals) | `print(10==20) -> False`, `print(10==10) -> True` |
| 4 | `!=` | not equal to | `print(10!=20) -> True`, `print(10!=10) -> False` |
| 5 | `>=` | greater than or equal to | `print(10>=20) -> False`, `print(10>=10) -> True` |
| 6 | `<=` | less than or equal to | `print(10<=20) -> True`, `print(10<=5) -> False` |

````python
print(10 > 5, 10 > 20)
print(10 < 5, 10 < 20)
print(10 == 20, 10 == 10)
print(10 != 20, 10 != 10)
print(10 >= 20, 10 >= 10)
print(10 <= 20, 10 <= 5)
````

```

>>> c = False > False

>>> b = False > False
>>> print(b)     -> False
```

````python
a = 10 > 20
print(a)

b = True >= False
print(b)

c = False > False
print(c)
````

## 4. Logical Operators (Comparison)

- The purpose of the logical operator is to compare the result of two or more relational expressions.
- If a logical operator connects two or more relational operators, then it is called a **logical operator expression** (compound test condition).
- The result of a logical expression is either `True` or `False`.
- In Python programming, we have **3 types** of logical operators. They are:

| S.No | Symbol | Meaning |
|---|---|---|
| 1 | `and` | Logical ANDing |
| 2 | `or` | Logical ORing |
| 3 | `not` | Logical NOT |

### 1) AND — `and` operator

**Syntax:** `RelExp1 and RelExp2`

The functionality of the `and` operator is expressed with the following truth table:

| RelExp1 | RelExp2 | RelExp1 and RelExp2 |
|---|---|---|
| True | False | False |
| False | True | False |
| False | False | False |
| True | True | True |

**Examples:**
```
>>> 10>2 and 20>10                     -> True
>>> 10>20 and 20>10                    -> False
>>> 10>20 and 20>10 and 30>50          -> False
>>> 10>5 and 30>40 and 40>20 and 20>10 -> False
>>> 10>2 and 4>2 and 4>1 and 40>20     -> True
```

````python
print(10 > 2 and 20 > 10)
print(10 > 20 and 20 > 10)
print(10 > 20 and 20 > 10 and 30 > 50)
print(10 > 5 and 30 > 40 and 40 > 20 and 20 > 10)
print(10 > 2 and 4 > 2 and 4 > 1 and 40 > 20)
````

### 2) OR — `or` operator

**Syntax:** `RelExp1 or RelExp2`

The functionality of the `or` operator is expressed with the following truth table:

| RelExp1 | RelExp2 | RelExp1 or RelExp2 |
|---|---|---|
| True | False | True |
| False | True | True |
| False | False | False |
| True | True | True |

**Examples:**
```
>>> 10>2 or 20>30                       -> True
>>> 10>20 or 40>20                      -> True
>>> 10>20 or 40>50 or 40>80             -> False
>>> 10<20 or 40>50 or 40>80             -> True
>>> 10>20 or 4>2 or 4>5 or 50>60        -> True
```

````python
print(10 > 2 or 20 > 30)
print(10 > 20 or 40 > 20)
print(10 > 20 or 40 > 50 or 40 > 80)
print(10 < 20 or 40 > 50 or 40 > 80)
print(10 > 20 or 4 > 2 or 4 > 5 or 50 > 60)
````

### 3) NOT — `not` operator

**Syntax:** `not RelExp`

The functionality of the `not` operator is expressed with the following truth table:

| RelExp1 | not RelExp1 |
|---|---|
| True | False |
| False | True |

**Examples:**
```
not True                 -> False
not False                -> True
not(10>20 and 20>30)     -> True
not(10<20 and 20<30)     -> False

not 10>2      -> False
not 10!=2     -> False
not 10==2     -> True
not 100       -> False
not 700       -> False
not 0         -> True
not 10-10     -> True
not "HYD"     -> False
not "0"       -> False
not "False"   -> False
not False     -> True
not [10, "TN", 23.45] -> False
not []        -> True
not {}        -> True
not ""        -> True
```

````python
print(not True)
print(not False)
print(not (10 > 20 and 20 > 30))
print(not (10 < 20 and 20 < 30))

print(not 10 > 2)
print(not 10 != 2)
print(not 10 == 2)
print(not 100)
print(not 0)
print(not (10 - 10))
print(not "HYD")
print(not "0")
print(not "False")
print(not False)
print(not [10, "TN", 23.45])
print(not [])
print(not {})
print(not "")
````

## Short Circuit Evaluation in the case of the `and` operator

If the `and` operator connects with multiple expressions (called logical expressions) and if the result of the initial relational expression is `False`, then the PVM will **not evaluate** the rest of the relational expressions, and the result of the entire logical expression is considered as `False`. This process of evaluation is called **Short Circuit Evaluation**.

## Short Circuit Evaluation in the case of the `or` operator

If the `or` operator connects with multiple relational expressions (called logical expressions) and if the result of the initial relational expression is `True`, then the PVM will **not evaluate** the rest of the relational expressions, and the result of the entire expression is considered `True`. This process of evaluation is also called **Short Circuit Evaluation**.

**Examples:**
```
100 or 200      -> 100
0 or 300        -> 300
100 or True     -> 100
```

````python
print(100 or 200)
print(0 or 300)
print(100 or True)
````

## Special points about Logical Operators

### `and` operator

`and` does not always return `True`/`False` — it returns one of the operands.

```
100 and 200            -> 200
100 and -234           -> -234
100 and 0              -> 0

100 and "False"        -> "False"

123 and 1.2            -> 1.2

```

````python
print(100 and 200)
print(100 and -234)
print(100 and 0)
print(0 and 120)
print(100 and True)
print(100 and "False")
print("True" and "False")
print(123 and 1.2)
print(100 and 200 and 300)
print(100 and 0 and 300)
print(0 and 0 and 0)
````

### `or` operator

`or` also returns one of the operands, not strictly `True`/`False`.

```

0 or 300             -> 300
100 or True          -> 100
0 or 0.5             -> 0.5
10 or 20 or 0        -> 10
0 or 10 or 0         -> 10
False or True        -> True
0 or "python" or -123 -> "python"

10 and 20 or 30              -> 20

100 and 200 and 300 or 400 and not 100 -> 300

True and "False" or True or "HYD" or "python"  -> "False"

```

````python
print(100 or 200)
print(0 or 300)
print(100 or True)
print(0 or 0.5)
print(10 or 20 or 0)
print(0 or 10 or 0)
print(False or True)
print(0 or "python" or -123)

print(10 and 20 or 30)
print(10 or 20 and -10 or 20)
print(100 and 200 and 300 or 400 and not 100)

print(True and "False" or True or "HYD" or "python")
print("" and "" or "")
````

## 5. Membership Operators

- The purpose of the membership operator is **"to check the existence of a particular value in an iterable object"**.
- An **iterable object** is one which contains more than one value (Ex: `str`, `bytes`, `bytearray`, `range`, `list`, `tuple`, `set`, `frozenset`, `dict`).
- A **non-iterable object** is one which contains a single value (`int`, `float`, `bool`, `complex`, `NoneType`).
- In Python programming, we have **2 types** of membership operators. They are:

1. `in`
2. `not in`

### 1. `in`

**Syntax:** `value in iterable-object`

Here the `in` operator returns `True` provided the `value` is present in the iterable-object.
Here the `in` operator returns `False` provided the `value` is not present in the iterable-object.

### 2. `not in`

**Syntax:** `value not in iterable-object`

Here the `not in` operator returns `True` provided the value is not present in the iterable-object.
Here the `not in` operator returns `False` provided the value is present in the iterable-object.

````python
s = "python"
print("p" in s)          # True
print("o" in s)          # True
print("p" not in s)      # False
print("Q" in s)          # False
print("pyt" in s)        # True
print("PYTON" in s)      # False
print("hon" in s)        # True
print("HON" in s)        # False
print("hyd" in s)        # False
````

**Example (slicing with membership operator):**
```
s = "PYTHON"
"PTO" in s                                -> False
"NOH" in s                                -> False
"NOH" in s[::-1]                          -> True
"PYT" not in s[::-1]                      -> True
"PTO" in s[::2][::-1]                     -> False
"PTO" in s[::2][::-1][::1]                -> True
"PTO" in s[-2::-2]                        -> True
"PTO"[::-1] not in s[-2:-2][::-1]         -> True
s in s                                    -> True
```

````python
s = "PYTHON"
print("PTO" in s)
print("NOH" in s)
print("NOH" in s[::-1])
print("PYT" not in s[::-1])
print("PTO" in s[::2][::-1])
print("PTO" in s[::2][::-1][::1])
print("PTO" in s[-2::-2])
print(s in s)
````

**Example 3 — membership on a list:**
```
lst = [100, "Reasom", 28.4, 21+3j]
"sum" in lst           -> False
"sum" in lst[1]        -> True
2 in lst[-1]           -> TypeError
2 in lst[-1].real      -> TypeError
```

````python
lst = [100, "Reasom", 28.4, 21 + 3j]
print("sum" in lst)
print("sum" in lst[1])
try:
    print(2 in lst[-1])
except TypeError as e:
    print("TypeError:", e)
````

**Example 4 — membership on a dict (checks keys):**
```
d1 = {10:"Apple", 20:"mango", 30:"kiwi"}

"Apple" not in d1                         -> True   (checks keys, not values)
10 in d1                                  -> True
20 not in d1                              -> False
d1[10][::-1] in d1.get(10)[::-1]          -> False
d1.get(100) not in d1                     -> True

s = "Apple is Red"
"e" in s                                  -> True
s[::-1] in s                              -> True

"1" in [1,2,3,4,7,8,9,0]                  -> False
"1" in str([1,2,4,5,6,7,8,9,0])           -> True
"1,2,3" not in str([])                    -> True
1,2,3 not in str([])                      -> TypeError
```

````python
d1 = {10: "Apple", 20: "mango", 30: "kiwi"}
print("Apple" not in d1)     # checks keys only
print(10 in d1)
print(20 not in d1)
print(d1.get(100) not in d1)

s = "Apple is Red"
print("e" in s)
print(s[::-1] in s)

print("1" in [1, 2, 3, 4, 7, 8, 9, 0])
print("1" in str([1, 2, 3, 4, 5, 6, 7, 8, 9, 0]))
print("1,2,3" not in str([]))
try:
    print((1, 2, 3) not in str([]))
except TypeError as e:
    print("TypeError:", e)
````

**More examples with strings/lists:**
```
1+2+3 not in str([1,2,3,4,5,6,7,8,9])        -> TypeError
str(1+2+3) not in str([1,2,3,4,5,6,7,8,9,0]) -> False
str(1+2+3) in str([1,2,3,4,5,6,7,8,9,0])     -> True

lst = ["Apple", "mango", 123, "True"]

```

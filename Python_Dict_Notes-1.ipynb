## 1. `dict` Category Data Type

- `dict` is a **pre-defined class**, and is treated as the `dict` data type.
- The purpose of the `dict` category data type is to store data in the form of **(key, value)** pairs.
- In `(key, value)`: the values of **key are always unique**, and the values of **value may or may not be unique**.
- To store `(key, value)` pairs in a `dict`, we use **curly braces `{ }`**, and entries must be separated by a **comma (`,`)**.

**Syntax:**
```
dictobj = {key1:val1, key2:val2, ..., keyN:valN}
```
- Here `key1, key2, ...` are values of *key*.
- `val1, val2, ...` are values of *value*.

- An object of `dict` maintains **insertion order**.
- An object of `dict` is **MUTABLE**.
- In the object of `dict`, the values of **key are IMMUTABLE** and values of **value are MUTABLE**.
- We have two types of `dict`:
  1. Empty dict
  2. Non-empty dict

### a) Empty dict
An empty dict is one which does not contain any elements, and whose length is zero.

**Syntax:**
```
dictobj = {}
or
dictobj = dict()
```

````python
dictobj = {}
print(dictobj, type(dictobj), len(dictobj))

dictobj2 = dict()
print(dictobj2, type(dictobj2), len(dictobj2))
````

### b) Non-Empty dict
A non-empty dict is one which contains elements, and whose length is > 0.

**Syntax:**
```
dictobj = {key1:val1, key2:val2, ..., keyN:valN}
```

**Syntax for adding (key, value) to a non-dict / empty dict:**
```
dictobj[key_n] = val_n     # where n = 1,2,3,...
```

#### Examples

````python
d1 = {10: 1.2, 20: 2.3, 30: 1.2, 40: 4.5}
print(d1, type(d1))
````

````python
d2 = {1: 'python', 2: 'c', 3: 'c++', 4: 'java'}
print(d2, type(d2))
````

````python
d3 = {10: 2, 10: 5, 10: 12}   # duplicate keys -> last one wins
print(d3)
````

````python
d1 = {100: 1.2, 200: 2.3, 300: 1.2, 400: 4.5}
print(d1, type(d1))
print(d1)
````

Accessing values by key (`keyerror` if the key does not exist):

````python
d1 = {100: 1.2, 200: 2.3, 300: 4.2, 400: 4.5}
print(d1[100])
print(d1[200])
print(d1[300])
print(d1[400])
````

````python
# accessing a key that does not exist raises KeyError
d1 = {100: 1.2, 200: 2.3, 300: 4.2, 400: 4.5}
try:
    print(d1[500])
except KeyError as e:
    print("KeyError:", e)
````

Modifying an existing entry:

````python
d1 = {100: 1.2, 200: 2.3, 300: 4.2, 400: 4.5}
d1[100] = 3.5     # modify entry
print(d1, id(d1))
print(len(d1))
````

Inserting new entries:

````python
d1 = {100: 1.2, 200: 2.3, 300: 4.2, 400: 4.5}
d1[0] = "Namdeo"     # inserted entry
d1[11] = "Ganga"
d1[12] = "Saraswati"
d1[13] = "Bhagirathi"
print(d1, type(d1), id(d1))
````

### 1. `clear()`
**Syntax:** `dictobj.clear()`

This function is used for removing all the elements of the `dict` object.

````python
d1 = {10: 10, 20: 'A', 30: 'B', 40: 'C'}
print(d1, type(d1), id(d1))

d1.clear()
print(d1)
print(len(d1))
````

````python
# calling clear() on an empty dict object -> still returns None, no error
print({}.clear())
print(dict().clear())
````

### 2. `pop(key)`
**Syntax:** `dictobj.pop(key)`

This function is used for removing `(key, value)` from a non-empty `dict` object by passing the value of key.

If the value of key does not exist then we get `KeyError`.

````python
d1 = {10: 1.2, 20: 3.4, 30: 4.5, 40: 2.3}
print(d1.pop(20))
print(d1.pop(30))
print(d1)
````

````python
d1 = {10: 1.2, 40: 2.3}
try:
    d1.pop(100)
except KeyError as e:
    print("KeyError:", e)

try:
    {}.pop(100)
except KeyError as e:
    print("KeyError:", e)
````

### 3. `popitem()`
**Syntax:** `dictobj.popitem()`

This function is used for removing the **last** `(key, value)` from a non-empty `dict` object.

When we call this function on an empty `dict` object, then we get `KeyError`.

````python
d1 = {10: 1.2, 20: 3.4, 30: 4.5}
print(d1, type(d1), id(d1))

print(d1.popitem())     # removes (30, 4.5) -> last inserted item
print(d1, id(d1))
````

````python
d1 = {10: 1.2, 20: 3.4}
print(d1.popitem())      # (20, 3.4)
print(d1.popitem())      # (10, 1.2)

try:
    d1.popitem()          # dict is empty now
except KeyError as e:
    print("KeyError:", e)

try:
    {}.popitem()
except KeyError as e:
    print("KeyError:", e)
````

### 4. `copy()` — Implements Shallow Copy
**Syntax:** `dictobj2 = dictobj1.copy()`

This function is used for copying the content of a `dict` object into another `dict` object.

````python
d1 = {10: 1.2, 20: 3.4, 30: 4.5, 40: 2.3}
print(d1, type(d1), id(d1))

d2 = d1.copy()          # shallow copy
print(d2, type(d2), id(d2))

d1[50] = 1.2             # inserted entry (only in d1)
d2[15] = 2.3             # inserted entry (only in d2)
print(d1, id(d1))
print(d2, id(d2))

d1[10] = 2.2              # modified entry (only in d1)
d2[10] = 7.8              # modified entry (only in d2)
print(d1, id(d1))
print(d2, id(d2))
````

### 5. `get()`
**Syntax:** `value = dictobj.get(key)`

This function is used for obtaining a value by passing the value of key from a non-empty `dict` object.

If the value of key does not exist then we get `None` as a result (no error).

````python
d1 = {10: "python", 20: 'c', 30: 'cpp', 40: 'django'}
print(d1)

val = d1.get(20)
print(val)

val = d1.get(30)
print(val)

val = d1.get(10)
print(val)

val = d1.get(130)
print(val)     # -> None (key does not exist)
````

````python
print({}.get(100))            # -> None
print(dict().get(100))        # -> None
````

**Alternative syntax:** `dictobj[key]`

This syntax also gives the value by passing the value of key. But if the value of key does not exist then we get `KeyError`.

````python
d1 = {10: 'py', 20: 'c', 30: 'cpp'}
print(d1)
print(d1[10])
print(d1[20])
print(d1[30])
try:
    print(d1[40])
except KeyError as e:
    print("KeyError:", e)
````

````python
# .get() on a nested dict value
d = {1: 1.2, 2: 2.2, 3: 3.3}
print(d.get(1))
print(d[2])
````

### 6. `keys()`
**Syntax:** `varname = dictobj.keys()`

This function is used for obtaining the values of *key* and placing them in `varname`, whose type is `<class 'dict_keys'>`.

````python
d1 = {10: 'python', 20: 'c', 30: 'cpp', 40: 'django'}
print(d1)

ks = d1.keys()
print(ks, type(ks))

for k in ks:
    print(k)
````

````python
for k in d1.keys():
    print(k)
````

### 7. `values()`
**Syntax:** `varname = dictobj.values()`

This function is used for obtaining the values of *value* and placing them in `varname`, whose type is `<class 'dict_values'>`.

````python
d1 = {10: 'py', 20: 'c', 30: 'cpp', 40: 'django'}
vs = d1.values()
print(vs, type(vs))

for v in vs:
    print(v)
````

````python
for v in d1.values():
    print(v)
````

### 8. `items()`
**Syntax:** `varname = dictobj.items()`

This function is used for obtaining `(key, value)` from a `dict` object in the tuples of a list and placing them in `varname`, whose type is `<class 'dict_items'>`.

````python
d1 = {10: 'py', 20: 'c', 30: 'cpp', 40: 'django'}
kvs = d1.items()
print(kvs, type(kvs))

for kv in kvs:
    print(kv)
````

````python
for kv in d1.items():
    print(kv)
````

````python
for k, v in d1.items():
    print(k, "->", v)
````

### 9. `update()`
**Syntax:** `dictobj1.update(dictobj2)`

This syntax is used for updating (insert / modify) the `dictobj1` `(key, value)` with `dictobj2` `(key, value)`.

`update()` is used for merging `dictobj1` with `dictobj2`.

````python
d1 = {10: 1.2, 20: 2.3}
d2 = {30: 1.2, 40: 1.4}
print(d1, id(d1))
print(d2, id(d2))

d1.update(d2)
print(d1, id(d1))    # d2's entries merged into d1
````

````python
# updating a common key modifies its value
d1 = {10: 1.2, 20: 2.3}
d2 = {20: 1.2, 40: 1.4}
d1.update(d2)
print(d1)     # value of key 20 is overwritten by d2's value
````

````python
d1 = {10: 1.2, 20: 2.3}
d2 = {1: 'py', 2: 'c'}
d3 = {'mango': 12.34, 'Apple': 4.5, 'kiwi': 20.00}

# update() accepts only ONE positional dict argument
try:
    d1.update(d2, d3)
except TypeError as e:
    print("TypeError:", e)
````

````python
d1 = {10: 1.2, 20: 2.3}
d2 = {1: 'py', 2: 'c'}
d3 = {'mango': 12.34, 'Apple': 4.5, 'kiwi': 20.00}

# update() returns None, so chaining .update() directly fails
try:
    d1.update(d2).update(d3)
except AttributeError as e:
    print("AttributeError:", e)
````

````python
# Solution: call update() separately (possible: dict + dict)
d1 = {10: 1.2, 20: 2.3}
d2 = {1: 'py', 2: 'c'}
d3 = {'mango': 12.34, 'Apple': 4.5, 'kiwi': 20.00}

d1.update(d2)
d1.update(d3)
print(d1)
````

````python
# dict + dict is NOT supported by the '+' operator
d1 = {10: 1.2}
d2 = {20: 2.3}
try:
    d3 = d1 + d2
except TypeError as e:
    print("TypeError:", e)
````

## 3. Combination — `dict` with `dict`, `set`, `tuple`, `list`

### Case 1: `dict` with `dict` → **possible** (nested dictionaries)

````python
d1 = {
    "sno": 10,
    "name": "RS",
    "Intmarks": {'C': 18, 'C++': 17, 'os': 19},
    "Extmarks": {'C': 70, 'C++': 65, 'os': 72},
    "cname": "excel"
}
print(d1)
````

````python
for k in d1:
    print(k, "->", d1.get(k))
````

````python
for k, v in d1.items():
    print(k, "->", v)
````

````python
# accessing / modifying nested dict
print(d1["Intmarks"]["C"])          # -> 18
d1["Intmarks"]["C"] = 20             # modify nested value
d1["Extmarks"]["DBMS"] = 65          # insert new nested (key,value)
print(d1)
````

````python
# removing an entry from a nested dict
d1["Extmarks"].pop("DBMS")           # remove ('DBMS', 65) nested entry
print(d1["Extmarks"])

d1.pop("Intmarks")                    # remove the whole 'Intmarks' entry
print(d1)
````

### Case 2: `dict` in `set` → **NOT possible** (`dict` is unhashable)

````python
d1 = {10: 'RS'}
try:
    s1 = {10, 'RS', d1}
except TypeError as e:
    print("TypeError:", e)
````

Because `dict` is neither *indexable nor hashable* — a `set` requires **hashable** elements.

### Case 3: `dict` in `tuple` → **possible**

````python
t1 = (10, "RS", {'C': 10, 'C++': 20}, "excel")
print(t1, type(t1))
````

### Case 4: `tuple` in `dict` → **possible** (as key and/or value)

````python
d1 = {
    "sno": 10,
    "name": "Surya",
    "Intmarks": (18, 20, 15),
    "cname": "excel"
}
for k, v in d1.items():
    print(k, "->", v, type(v))
````

````python
# a tuple value can be sorted and re-assigned
d1["Intmarks"] = tuple(sorted(d1["Intmarks"]))
for k, v in d1.items():
    print(k, "->", v)
````

````python
d1["Extmarks"] = (34, 56, 78)
for k, v in d1.items():
    print(k, "->", v)
````

````python
print(d1["Extmarks"][1])           # -> 56
print(d1["Extmarks"][-1])          # -> 78
print(d1["Extmarks"][len(d1["Extmarks"]) - 1])   # -> 78
````

### Case 5: `set` in `dict` → **possible** (as a value)

````python
d1 = {"sno": 10, "name": "Surya", "Intmarks": {18, 20, 15}, "cname": "excel"}
print(d1["Intmarks"], type(d1["Intmarks"]))

d1["Intmarks"].add(125)
for k, v in d1.items():
    print(k, "->", v, type(v))
````

### Case 6: `list` in `dict` → **possible** (as a value)

````python
d1 = {"sno": 10, "name": "Surya", "Intmarks": [15, 20, 18], "cname": "excel"}

for k, v in d1.items():
    print(k, "->", v, type(v))
````

````python
for val in d1["Intmarks"]:
    print(val)
````

````python
d1["Intmarks"].insert(1, 18)     # insert into the nested list
for val in d1["Intmarks"]:
    print(val)
print(d1["Intmarks"])
````

### Case 7: `dict` in `list` → **possible**

````python
lst = [100, 'RS', {"TS": "HYD", "KAR": "Bang", "MH": "mum"}, "excel"]

for val in lst:
    print(val, type(val), type(lst))
````

````python
for k in lst[2].keys():
    print(k, "->", lst[2].get(k))
````

````python
lst[2]["AP"] = "VIJ-VIJAG"           # insert into nested dict inside list
print(lst)
````

````python
lst.insert(2, {'one': 1, 'two': 2, 'Three': 3})   # insert a new dict into the list
print(lst)

for k, v in lst[2].items():
    print(k, "->", v)
````

## 4. `NoneType` Category Data Type

- `None` is the **keyword**, and is treated as the value of `NoneType` data type.
- `NoneType` is a pre-defined class.
- `None` is **not** `False`, `""` (empty string), `0`, and `None` **is** nothing (represents absence of value).
- We can't create an object of `NoneType` because it contains a single value.

````python
a = None
print(a, type(a))
````

````python
print(None == 0)
print(None == "")
print(None == False)
````

````python
print(None is None)
````

````python
try:
    n = NoneType()
except NameError as e:
    print("NameError:", e)     # 'NoneType' is not defined in the current namespace
````

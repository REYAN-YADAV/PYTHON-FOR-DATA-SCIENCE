## 1. What is a Loop?

The purpose of looping (iterative / repetitive) statements is to perform a certain operation repeatedly for a finite number of times, until the test condition becomes **False**.

In Python we have **2 types** of looping / iterative / repetitive statements:

1. `while` loop, or `while...else` loop
2. `for` loop, or `for...else` loop

While solving any problem using a loop, we must ensure **3 important parts**:

1. **Initialization part** — makes us understand where to start.
2. **Test condition** — makes us understand where to stop (loop runs until the condition becomes `False`).
3. **Updation part** — makes us understand how to move forward or backward (increment / decrement).


````python
# Example: generate 1 to 10 numbers
i = 1
while i <= 10:
    print(i)
    i = i + 1
````

## 2. `while` Loop / `while...else` Loop

**Syntax 1 (simple while):**
```python
while (test_cond):
    statement-1
    statement-2
    ...
    statement-n
other statement in program
```

**Syntax 2 (while...else):**
```python
while (test_cond):
    statement-1
    statement-2
    ...
    statement-n
else:
    else block of statement   # while body
other statement in program
```

- Here `while` and `else` are **keywords**.
- `test_cond` can be either `True` or `False`.
- If `test_cond` is `True`, then Python executes the indented block of statements, and control goes back to `test_cond`. If it is `True` again, Python executes the indented block again. This process repeats until `test_cond` becomes `False`.
- Once `test_cond` becomes `False`, Python executes the `else` block of statements (if present) and then executes other statements in the program.


## 3. `for` Loop / `for...else` Loop

**Syntax 1:**
```python
for varname in Iterable_Object:
    Block statements
other statements in program
```

**Syntax 2 (for...else):**
```python
for varname in Iterable_Object:
    Block statements
else:
    Else block of statements
other statements in program
```

### Explanation
- `for`, `in`, and `else` are **keywords**.
- `Iterable_Object` can be a sequence type (`bytes`, `bytearray`, `range`, `str`) or `list`, `tuple`, `set`, `frozenset`, `dict`.
- Execution: each element of the iterable object is selected, placed into `varname`, and the indented block of statements is executed. This repeats until all elements of the iterable object are exhausted.
- When all elements are exhausted, Python control comes out of the loop and executes the `else` block of statements (if written) written under the `else` block.
- Writing the `else` block is **optional**.

### Q1. Generate 1 to n numbers, where n is a positive number (using `while` loop)

If `n <= 0`, print an *Invalid input* message; otherwise generate the numbers 1 to n. (Note from the original notes: if you don't initialize `i`, you get a `NameError`. If you don't write the `while...else` test condition's `else`, no error is shown — Python just runs the other statements directly.)


````python
def generate_1_to_n(n):
    if n <= 0:
        print(f"Invalid input: {n}")
        print("Please input a positive number")
    else:
        i = 1
        while i <= n:
            print(i)
            i = i + 1
        else:
            print("Thank you")

generate_1_to_n(10)
````

### Q2. Generate n to 1 numbers, where n is a positive integer (using `while` loop)


````python
def generate_n_to_1(n):
    if n <= 0:
        print(f"Invalid input: {n}")
        print("Please input a positive number")
    else:
        i = n
        while i >= 1:
            print(i)
            i = i - 1
        else:
            print("Thank you")

generate_n_to_1(10)
````

### Q4. Generate even numbers from n in both directions (ascending and descending), where n is a positive integer


````python
def even_both_directions(n):
    print("Ascending:")
    i = 2
    while i <= n:
        print(i)
        i = i + 2
    print("Descending:")
    i = n if n % 2 == 0 else n - 1
    while i >= 2:
        print(i)
        i = i - 2

even_both_directions(20)
````

### Q5. Accept any word/line of text, display each character, and also display it in reverse (using `while` loop and indexing)


````python
def chars_forward_and_reverse(line):
    print("Characters (forward):")
    i = 0
    while i <= len(line) - 1:
        print(line[i])
        i = i + 1
    print("Characters (reverse):")
    i = -1
    while i >= -len(line):
        print(line[i])
        i = i - 1

chars_forward_and_reverse("abc")
````

### Q6. Find the sum of n natural numbers, where n is positive (using `while` loop)


````python
def sum_of_n_natural(n):
    total = 0
    i = 1
    while i <= n:
        total = total + i
        i = i + 1
    print(f"Sum of first {n} natural numbers = {total}")
    print(f"Using formula n(n+1)/2 = {n * (n + 1) // 2}")

sum_of_n_natural(5)
````

### Q7. Find the sum of squares of n natural numbers (using `while` loop)


````python
def sum_of_squares(n):
    total = 0
    i = 1
    while i <= n:
        total = total + i ** 2
        i = i + 1
    print(f"Sum of squares of first {n} natural numbers = {total}")

sum_of_squares(5)
````

### Q9. Calculate the factorial of a given number (using `while` loop)


````python
def factorial(n):
    fact = 1
    i = 1
    while i <= n:
        fact = fact * i
        i = i + 1
    print(f"Factorial of {n} = {fact}")

factorial(5)
````

### Q13. Accept a numerical value and find the sum of its digits

Example: `1256` &rarr; `1 + 2 + 5 + 6`

- **Test case 1:** the given value must be a number.
- **Test case 2:** the given value should not be a string with alphabets/symbols.


````python
def sum_of_digits(value):
    if not str(value).isdigit():
        print("Invalid input: value must be a positive number "
              "(no alphabets or symbols allowed)")
        return
    n = int(value)
    total = 0
    while n > 0:
        total = total + (n % 10)
        n = n // 10
    print(f"Sum of digits of {value} = {total}")

sum_of_digits(1256)    # valid
sum_of_digits("12a5")  # invalid
````

### Q14. Accept any value and print it in reverse **without** using the slice operation (`[::-1]`)


````python
def reverse_without_slice(value):
    rv = ""
    for index in range(len(value) - 1, -1, -1):
        rv = rv + value[index]
    else:
        print(f"Reversed value = {rv}")
    return rv

reverse_without_slice("python")
````

### Q16. Accept a line of text and find each word and its length

Input: `python is an oop lang`

Expected output: `python -> 6`, `is -> 2`, `oop -> 3`, `lang -> 4`


````python
def word_lengths(text):
    words = text.split()
    d = {}
    for word in words:
        d[word] = len(word)
    for k, v in d.items():
        print(f"{k} -> {v}")
    return d

word_lengths("python is an oop lang")
````

### Q17. Accept a line of text and find the highest length word

Input: `python and oop lang` &rarr; `python -> 6, and -> 3, oop -> 3, lang -> 4`

Expected output: `Highest length word = python`


````python
def highest_length_word(text):
    words = text.split()
    d = {}
    for word in words:
        d[word] = len(word)
    else:
        dv = max(list(d.values()))
        lst = list()
        for k, v in d.items():
            if v == dv:
                lst.append(k)
        else:
            print("Highest Length word(s):")
            for val in lst:
                print("|*|{}".format(val))

highest_length_word("python and oop lang")
````

### Q18. Accept any number and find all of its factors

Input: `24` &rarr; Expected output: `1, 2, 3, 4, 6, 8, 12, 24`

Input: `50` &rarr; Expected output: `1, 2, 5, 10, 25, 50`


````python
def find_factors(n):
    factors = []
    for i in range(1, (n // 2) + 1):
        if n % i == 0:
            factors.append(i)
    factors.append(n)
    print(f"Factors of {n} = {factors}")
    return factors

find_factors(24)
find_factors(50)
````

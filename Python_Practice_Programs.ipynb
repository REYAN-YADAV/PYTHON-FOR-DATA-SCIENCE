## Q. Palindrome or Not

**Input:** `LIRIL`
**Expected:** Palindrome (because reverse of `'LIRIL'` == `'LIRIL'`) -> `True`

**Input:** `python`
**Expected output:** Not palindrome (because reverse of `'python'` == `'nohtyp'` != `'python'`) -> `False`

```python
word = input("Enter a word: ")
res = "palindrome" if word == word[::-1] else "Not palindrome"
print("'{}' is {}".format(word, res))
```

````python
def check_palindrome(word):
    res = "palindrome" if word == word[::-1] else "Not palindrome"
    print("'{}' is {}".format(word, res))

check_palindrome("LIRIL")
check_palindrome("python")
````

### Case-insensitive version

A word like `"Liril"` isn't equal to its own reverse (`"lirIL"`) as a raw string comparison,
because the letter casing doesn't mirror. Converting both sides to lower case first fixes this:

```python
res = "palindrome" if word.lower() == word[::-1].lower() else "Not palindrome"
print("'{}' is {}".format(word, res))
```

````python
def check_palindrome_ci(word):
    res = "palindrome" if word.lower() == word[::-1].lower() else "Not palindrome"
    print("'{}' is {}".format(word, res))

check_palindrome_ci("LIRIL")
check_palindrome_ci("Liril")
check_palindrome_ci("python")
check_palindrome_ci("Malayalam")
````

## Q. Write a Python program which will accept 3 numerical values, find the bigger among them, and check for equality

**Test case 1**

```python
try:
    a = float(input("Enter value of a: "))
    b = float(input("Enter value of b: "))
    c = float(input("Enter value of c: "))

    res = (a if (a > b) and (a > c)
           else b if (b > a) and (b > c)
           else c if (c > a) and (c > b)
           else "All values are equal")

    print("Big({},{},{}) = {}".format(a, b, c, res))
except ValueError:
    print("Don't enter alnum, str, symbol")
```

**Note:** the strict `>` comparisons above correctly report `"All values are equal"` only
when **all three** values are equal. If exactly two values tie for the largest (e.g. `10, 10, 5`),
none of the three strict conditions is satisfied, so this version would also fall through to
`"All values are equal"` — which is misleading. A version using `>=` handles ties for the maximum
more precisely:

```python
res = (a if a >= b and a >= c
       else b if b >= a and b >= c
       else c if c >= a and c >= b
       else "All values are equal")
```

````python
def find_big3(a_str, b_str, c_str):
    try:
        a = float(a_str)
        b = float(b_str)
        c = float(c_str)

        res = (a if (a > b) and (a > c)
               else b if (b > a) and (b > c)
               else c if (c > a) and (c > b)
               else "All values are equal")

        print("Big({},{},{}) = {}".format(a, b, c, res))
    except ValueError:
        print("Don't enter alnum, str, symbol")

find_big3("10", "20", "30")
find_big3("50", "20", "30")
find_big3("10", "10", "10")
find_big3("abc", "20", "30")
````

````python
def find_max3(a_str, b_str, c_str):
    try:
        a = float(a_str)
        b = float(b_str)
        c = float(c_str)

        res = (a if a >= b and a >= c
               else b if b >= a and b >= c
               else c if c >= a and c >= b
               else "All values are equal")

        print("Big({},{},{}) = {}".format(a, b, c, res))
    except ValueError:
        print("Don't enter alnum, str, symbol")

find_max3("10", "10", "5")   # tie for max, handled correctly with >=
find_max3("10", "20", "30")
find_max3("7", "7", "7")
````

## Q. Write a Python program which accepts any word and decides whether it is a vowel or not

```python
word = input("Enter any word: ")
res = "vowel" if word.lower() in "aeiou" else "not a vowel"
print("{} is {}".format(word, res))
```

**Note:** `word.lower() in "aeiou"` correctly checks *single-character* vowels
(any of `a`, `e`, `i`, `o`, `u`, regardless of case). For a **multi-character**
input, `in` would instead check whether the whole word appears as a
*substring* of `"aeiou"` (almost always `False`) — so this check is intended
for single letters. A version that also handles full words is shown below.

````python
def check_vowel(word):
    res = "vowel" if word.lower() in "aeiou" else "not a vowel"
    print("{} is {}".format(word, res))

check_vowel("A")
check_vowel("e")
check_vowel("b")
check_vowel("U")
````

````python
def check_vowel_word(word):
    # Works for a single letter OR checks the first letter of a full word
    first = word[0].lower() if word else ""
    res = "vowel" if first in "aeiou" else "not a vowel"
    print("{} is {}".format(word, res))

check_vowel_word("Apple")
check_vowel_word("Banana")
check_vowel_word("Orange")
check_vowel_word("Mango")
````

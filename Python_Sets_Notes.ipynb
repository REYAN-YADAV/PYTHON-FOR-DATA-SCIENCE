## 1. Set Category Data Types

**Set Theory** → Collection of well-defined, unique elements (values).

**Example**
```
A = {10, 20, 30, 40}
B = {15, 20, 30, 35}
Find A ∪ B
A ∪ B = {10, 20, 30, 40, 15, 35}
```

**Purpose of Set Category Data Type**

The purpose of the set category data type is to store multiple values, either of the
same type, different types, or both types, in a single object with **unique values**
(no duplicates are allowed).

In Python programming, we have 2 types of Data Types under the **Set category**:

1. `set` — mutable
2. `frozenset` — immutable


## 2. Properties of Set

1. `set` is one of the pre-defined classes and is treated as a **set data type**.

2. The purpose of the set data type is to store multiple values, either of the same
   type or different types (or both), in a single object with unique values
   (**no duplicates are allowed**).

3. The elements/values of a set data type must be written/stored inside **curly
   brackets `{ }`**.

   **Syntax:** `objname = {val1, val2, ..., valn}`

4. An object of set never maintains **insertion order** — PVM (Python Virtual
   Machine) can display the elements of a set in **any one possibility**.

5. On the object of set, we **cannot perform indexing and slicing** operations,
   since a set neither maintains insertion order nor supports indexing.

6. An object of set belongs to **both**:
   - **Immutable** in the sense that a set object does not support *item
     assignment*.
   - **Mutable** in the case of adding, deleting, modifying elements, etc.


````python
A = {10, 20, 30, 40}
B = {15, 20, 30, 35}
print(A | B)          # union operator
````

````python
# Duplicates are automatically removed
s1 = {10, 20, 30, 40, 50, 60, 10, 20}
print(s1)
````

````python
s1 = {10, 20, 30, 40}
print(s1[0])           # TypeError: 'set' object is not subscriptable
````

````python
s2 = {10, 45, 'python', True, 'Regd'}
print(s2, type(s2), id(s2))
````

````python
s2[0] = 'Django'      # TypeError: 'set' object does not support item assignment
````

````python
s2.add('Django')       # allowed - adding is a mutable operation
print(s2)
````

## 3. Types of Set Objects

In Python programming, we can create **two** types of set objects:

1. **Empty Set**
2. **Non-Empty Set**


### 3.1 Empty Set

An **empty set** is one which does not contain any elements and whose length is `0`.

**Syntax:** `varname = set()`


````python
s1 = set()
print(s1, type(s1))
print(len(s1))
````

### 3.2 Non-Empty Set

A **non-empty set** is one which contains elements, and whose length is `> 0`.

**Syntax:**
```
varname = {val1, val2, ..., valn}
    or
varname = set(object)
```


````python
s2 = set('shiny')                 # iterable -> set of characters
print(s2, type(s2))

s3 = {10, 'shiny', 'Node', 3.33}
print(s3, type(s3))
print(len(s3))
````

### 4.1 `clear()`

Used for removing all elements from a set object.

**Syntax:** `setobj.clear()`

When we call this function on an empty set object, we get `None` (no error) as a
result — since there's nothing to clear, the operation simply completes.


````python
s1 = {10, 20, 30, 'python'}
print(s1, type(s1), id(s1))
print(len(s1))

s1.clear()
print(s1, type(s1), id(s1))

print(set().clear())      # None
````

### 4.2 `add()`

Used for adding a value to a set object.

**Syntax:** `setobj.add(value)`


````python
s1 = {10, 'shy'}
print(s1, type(s1), id(s1))

s1.add('pg')
s1.add(23.45)
print(s1, type(s1), id(s1))

s2 = set()
print(s2, type(s2), id(s2))

s2.add(100)
s2.add('py')
s2.add('Ross')
print(s2, type(s2), id(s2))
````

### 4.3 `remove()`

Used for removing the specified value from a set object.

**Syntax:** `setobj.remove(value)`

If the specified value doesn't exist in the set object, then we get a `KeyError`.


````python
s1 = {10, 20, 30, 40, 10, 55, 65, 75}
print(s1, type(s1), id(s1))

s1.remove(10)
print(s1, id(s1))

s1.remove(40)
print(s1)

s1.remove(140)          # KeyError: 140

set().remove(40)        # KeyError: 40
````

### 4.4 `discard()`

Used for removing the specified value from a set object.

**Syntax:** `setobj.discard(value)`

If the specified value does not exist in the set object, then we simply get **no
output / `None`** — unlike `remove()`, `discard()` does NOT raise an error.


````python
s1 = {10, 20, 30, 40, 11, 55, 65, 75}
print(s1, type(s1), id(s1))

s1.discard(10)
print(s1, id(s1))                 # {20, 30, 40, 55, 65, 75}

s1.discard(30)
print(s1, id(s1))                 # {20, 40, 55, 65, 75}

s1.discard(130)                   # No output / None

s1.remove(130)                    # KeyError: 130

print(set().discard(1))           # None
print(set().remove(1))            # KeyError
````

### 4.5 `pop()`

Used for removing/retrieving an element without order of display — removes any
**arbitrary** element from the set.

**Syntax:** `setobj.pop()`

If you use `print()` to display, then...


````python
s1 = {10, 20, 30, 50, 60}
s1.pop()          # removes an arbitrary element (order not guaranteed)
print(s1.pop())
print(s1.pop())

s1.pop()
s1.pop()
s1.pop()           # KeyError: 'pop from an empty set'

set().pop()         # KeyError
````

````python
# Example: given order of display -> remove first element
s2 = {10, 'shiny', 44.4, 'py', True, 2 + 3j}
print(s2, type(s2), id(s2))

s2.pop()
s2.pop()
s2.pop()
s2.pop()
s2.pop()
s2.pop()
s2.pop()           # KeyError: pop from an empty set
````

### 4.6 `copy()`

**Implementation of shallow copy** — used for copying the contents of one set
object to another set object.

**Syntax:** `setobj2 = setobj1.copy()`


````python
s1 = {10, 20, 30, 40}
print(s1, type(s1), id(s1))

s2 = s1.copy()          # shallow copy - different identity, same contents
print(s2, type(s2), id(s2))

s1.add(100)
s2.add(400)

print(s1)               # {10, 100, 20, 30, 40}
print(s2)               # {400, 10, 20, 30, 40}
````

### 4.7 `isdisjoint()`

**Syntax:** `setobj1.isdisjoint(setobj2)`

This function returns `True`, provided there is **no common element** in both
`setobj1` and `setobj2`.

This function returns `False`, provided there is at least **one common element** in
both `setobj1` and `setobj2`.


````python
s1 = {10, 20, 30}
s2 = {15, 45, 35}
s3 = {15, 45, 55}

print(s1.isdisjoint(s2))     # True
print(s2.isdisjoint(s3))     # False

print(set().isdisjoint(s1))  # True
print(set().isdisjoint(set()))   # True
````

### 4.8 `issuperset()`

**Syntax:** `setobj1.issuperset(setobj2)`

This function returns `True`, provided `setobj1` contains **all the elements** of
`setobj2`. i.e. all the elements of `setobj2` are present in `setobj1`.


````python
s1 = {10, 20, 30, 40}
s2 = {10, 20}
s3 = {10, 20, 50, 60}

print(s1.issuperset(s2))     # True
print(s2.issuperset(s1))     # False
print(s1.issuperset(s3))     # False
print(s3.issuperset(s1))     # False
print(s3.issuperset(s2))     # True
print(s2.issuperset(s3))     # False

print(s1.issuperset(set()))  # True   -> every set is a superset of the empty set
print(set().issuperset(set()))  # True
````

### 4.9 `issubset()`

**Syntax:** `setobj1.issubset(setobj2)`

This function returns `True`, provided **all the elements of `setobj1`** are present
in `setobj2`, i.e. `setobj2` contains all the elements of `setobj1`.

This function returns `False`, provided **at least one element of `setobj1`** is
NOT present in `setobj2`.


````python
s1 = {10, 20, 30, 40}
s2 = {10, 20}
s3 = {10, 20, 30, 40}

print(s2.issubset(s1))     # True
print(s2.issubset(s3))     # True
print(s1.issubset(s2))     # False
print(s3.issubset(s2))     # False

print(set().issubset(s1))          # True
print(s1.issubset(set()))          # False
print(s1.issubset(s1))             # True
print(set().issubset(set()))       # True

print(set({10, 20, 30}).issubset(set({10, 20, 30})))   # True
print(set('PYTHON').issubset(set('python')))            # False
````

### 4.10 `union()`

**Syntax:** `setobj3 = setobj1.union(setobj2)`

This function is used for obtaining all the **unique elements** of `setobj1` and
`setobj2`, and the resultant elements are placed in `setobj3`.


````python
s1 = {10, 20, 30}
s2 = {20, 40, 50}

s3 = s1.union(s2)
print(s3)             # {10, 20, 30, 40, 50}

s4 = s2.union(s1)
print(s4)             # {10, 20, 30, 40, 50}
````

### 4.11 `intersection()`

**Syntax:** `setobj3 = setobj1.intersection(setobj2)`

This function is used to obtain all the **common elements** of `setobj1` and
`setobj2`, and the resultant common elements are placed in `setobj3`.


````python
s1 = {10, 20, 30}
s2 = {20, 40, 50}

s3 = s1.intersection(s2)
print(s3)              # {20}

s4 = s2.intersection(s1)
print(s4)              # {20}
````

### 4.12 `difference()`

**Syntax:** `setobj3 = setobj1.difference(setobj2)`

This function is used for removing the common elements from `setobj1` and
`setobj2`, and takes the remaining elements of `setobj1` and places them in
`setobj3`.


````python
s1 = {10, 20, 30}
s2 = {20, 40, 50}

s3 = s1.difference(s2)
print(s3)              # {10, 30}

s3 = s2.difference(s1)
print(s3)              # {40, 50}
````

**By formula:**
```
s1 = {10, 20, 30}
s2 = {20, 40, 50}
s3 = s1.union(s2).difference(s1.intersection(s2))
print(s3)     # {10, 30, 40, 50}
```


````python
s1 = {10, 20, 30}
s2 = {20, 40, 50}
s3 = s1.union(s2).difference(s1.intersection(s2))
print(s3)
````

### 4.13 `symmetric_difference()`

`S1 Δ S2 = S1 ∪ S2 − S1 ∩ S2`

In maths, this function is called the **Delta Operator** — common elements are
removed and the remaining (uncommon) elements from both sets are taken.

**Syntax:**
```
setobj3 = setobj1.symmetric_difference(setobj2)
    or
setobj3 = setobj1.union(setobj2).difference(setobj1.intersection(setobj2))
```

This function is used for removing the common elements from both `setobj1` and
`setobj2`, and takes the remaining elements from both `setobj1` and `setobj2`, and
places them in `setobj3`.


````python
s1 = {10, 20, 30}
s2 = {20, 40, 50}

s3 = s1.symmetric_difference(s2)
print(s3)              # {10, 30, 40, 50}

s3 = s2.symmetric_difference(s1)
print(s3)              # {10, 30, 40, 50}
````

### 4.14 `symmetric_difference_update()`

**Syntax:** `setobj1.symmetric_difference_update(setobj2)`

This function is used for removing the common elements from both `setobj1` and
`setobj2`, and takes the remaining elements from `setobj1` and `setobj2`, and
places them **back into `setobj1`** (modifies `setobj1` in place; returns `None`).


````python
s1 = {10, 20, 30}
s2 = {20, 25, 35}

print(s1, id(s1))
print(s2, id(s2))

s1.symmetric_difference_update(s2)
print(s1, id(s1))         # {10, 25, 30, 35} -- same id, modified in place
````

### 4.15 `update()`

Used for adding or updating (modifying) the content of `setobj1` with `setobj2`.
Since `set` is a mutable object type in Python, this is possible (unlike a
`frozenset`, which is immutable).

**Syntax:** `setobj1.update(setobj2)`


````python
s1 = {10, 20, 30}
s2 = {15, 25, 35}

s1.update(s2)
print(s1)          # {10, 15, 20, 25, 30, 35}
````

**Note:** the same functionality is exposed through
`symmetric_difference_update`, `update`, `intersection_update`,
`difference_update` — each modifies the object of `set` in place, since `set` is
mutable.


````python
s1 = {10, 20, 30}
s2 = {10, 20, 30}

s1.update(s2)
print(s1)          # {10, 20, 30} (no new/unique elements from s2)
````

## 5. Inner Sets / Nested Sets

Since a `set` in Python is an **un-hashable type**, no `set` can be placed directly
inside another `set` (or as a dictionary key).

1. **`set` in `set`** → **Not possible** (un-hashable type)
2. **`set` in `list`** → Possible
3. **`set` in `tuple`** → Possible

4. **`list` in `list`** → Possible
5. **`list` in `tuple`** → Possible
6. **`list` in `set`** → **Not possible** (un-hashable type; `list` is mutable)

> `un-hashable` means: no indexing is possible, nor is updation possible.

7. **`tuple` in `tuple`** → Possible
8. **`tuple` in `list`** → Possible
9. **`tuple` in `set`** → **Possible**, because one can write an immutable object
   (`tuple`) inside another immutable/hashable-compatible object (`set`).


````python
s1 = {10, 'RS', {11, 12}, 25}          # TypeError: unhashable type: 'set'
````

````python
s2 = {10, 'RS', [11, 12], 25}          # TypeError: unhashable type: 'list'
````

````python
s3 = {10, 'RS', (11, 12), 25}          # OK - tuple is hashable
print(s3)
````

````python
l1 = [10, 'RS', {11, 12}, {56, 78}, 25]
print(l1)

tl = (10, 'RS', {11, 12}, {56, 78}, 25)
print(tl)
````

## 6. Practice Problem — Cricket & Tennis Players

```
Set of cricket players = {'Kohli', 'Sachin', 'Rohit'}
Set of tennis players  = {'Rossum', 'Travis', 'Rohit'}
```

**Venn diagram:** the two circles ("Cricket" and "Tennis") sit inside a universal
set box; `Rohit` lies in the overlap.

**Questions**
1. Find all the players who are playing either cricket or tennis or both.
2. Find all the players who are playing both cricket and tennis.
3. Find all the players who are playing only cricket.
4. Find all the players who are playing only tennis.
5. Find all the players who are playing exclusively cricket and tennis (i.e. not
   both).


````python
cp = {'Rohit', 'Kohli', 'Sachin'}
tp = {'Rossum', 'Travis', 'Rohit'}

# Q1: playing either cricket or tennis or both  -> union
all_cp_tp = cp.union(tp)
print(all_cp_tp)
# {'Rohit', 'Rossum', 'Travis', 'Kohli', 'Sachin'}
````

````python
# Q2: playing both cricket and tennis -> intersection
both_cp_tp = cp.intersection(tp)
print(both_cp_tp)
# {'Rohit'}
````

````python
# Q3: playing only cricket -> cp - tp
only_cp = cp.difference(tp)
print(only_cp)
# {'Kohli', 'Sachin'}
````

````python
# Q4: playing only tennis -> tp - cp
only_tp = tp.difference(cp)
print(only_tp)
# {'Rossum', 'Travis'}
````

````python
# Q5: playing exclusively cricket and tennis (not both) -> symmetric_difference
excl_cp_tp = cp.symmetric_difference(tp)
print(excl_cp_tp)
# {'Sachin', 'Rossum', 'Ross', 'Travis'}
````

### Alternate solutions using operators

`|` = union, `&` = intersection, `-` = difference, `^` = symmetric difference


````python
cp = {'Rohit', 'Kohli', 'Sachin'}
tp = {'Rossum', 'Travis', 'Rohit'}

print(cp | tp)     # Q1: union
print(cp & tp)     # Q2: intersection
print(cp - tp)     # Q3: only cricket
print(tp - cp)     # Q4: only tennis
print(cp ^ tp)     # Q5: exclusive (symmetric difference)
````

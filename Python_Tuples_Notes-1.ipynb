# List Category Data Types
### (Collection Data Type / Data Structure)

The purpose of a **List category** data type is to store multiple values, either of the *same type* or *different type* (or both), in a single object — with support for both unique and duplicate members.

We have **two** data types in the List category:

1. **List** — mutable
2. **Tuple** — immutable

## Properties of Tuple

- Tuple is one of the pre-defined classes and is treated as a List data type.
- **Constant** → a constant value that never changes during the execution of a program.
- The purpose of a tuple is to store multiple values, either of the same type, different type, or both types.
- In other words, **tuple is always used for maintaining constant values.**
- To store data in a tuple object, the data must be written / enclosed within **braces `()`**.

**Syntax:**
```
varname = (val1, val2, val3, ... val-n)
```

- An object of tuple **maintains insertion order**.
- On a tuple object, we can perform **both indexing and slicing** operations.
- An object of tuple belongs to **immutable**, so a tuple object **does not support item assignment**.

In Python, we can create **two types** of tuple:
1. Empty tuple
2. Non-Empty tuple


### 1. Empty Tuple
An empty tuple is one which does **not** contain any value (element) and whose length is **0**.

**Syntax:**
```
varname = ()
varname = tuple()
```

### 2. Non-Empty Tuple
A non-empty tuple is one which contains elements and whose length is **not equal to 0**.

**Syntax:**
```
varname = (val1, val2, ... val-n)
varname = tuple(object)
```

**Example:**


````python
t1 = (10, 20, 30, 40)
print(t1, type(t1))
print(t1[0], t1[1], t1[-1], t1[-4])
````

Tuples do **not support item assignment** since they are immutable:
```python
t1 = (10, 20, 30, 40)
t1[0] = 10
```
```
TypeError: 'tuple' object does not support item assignment
```


### Converting a string into a tuple

````python
s = "python"
t = tuple(s)
print(t)
````

### Tuple without parentheses
Values separated only by commas are also treated as a tuple:

````python
t2 = 10, "Rossom", 44.4, 'NL'
print(t2, type(t2))
````

### Converting between list and tuple

````python
l2 = list(t2)
print(l2, type(l2))

t3 = tuple(l2)
print(t3, type(t3))
````

## Pre-defined Functions Present in Tuple

We already know that on a tuple object we can perform both indexing and slicing operations. Along with these, tuple also provides the following pre-defined functions:

1. `index()`
2. `count()`


````python
t1 = (10, "RS", 45.67)
print(t1.index(10))
print(t1.index('RS'))
print(t1.count(10))
print(t1.count(100))
````

### `id()` and copying a tuple

By assigning `t2 = t1`, only a **deep copy reference is possible, not a shallow copy** — both names point to the **same object** (same `id`).


````python
t1 = (10, 20, 30, 40, 50, 10)
print(t1, id(t1), type(t1))

t2 = t1   # not a new object, same reference
print(t2, id(t2), type(t2))
print(t1 is t2)
````

**Note:** By using the `del` operator, we **cannot** delete a value of a tuple object using indexing or slicing, but we **can** delete the entire object, because a tuple object belongs to the immutable category.

```python
del t1[0]        # TypeError: 'tuple' object doesn't support item deletion
del t1            # This works - removes the complete object
print(t1)         # NameError: name 't1' is not defined
```


## `sorted()` Function

This function is used for **sorting** the data of an immutable tuple object, and it gives back the sorted data in the form of a **list**.

**Syntax:**
```
list = sorted(tuple_object)
```

**Example:**


````python
t1 = (10, 23, -56, -1, 13, 15, 10)

# t1.sort()  -> AttributeError: 'tuple' object has no attribute 'sort'

x = sorted(t1)
print(x, type(x))
````

We can convert the sorted list back into a tuple:

````python
t1 = tuple(x)   # converted sorted list into tuple
print(t1, type(t1))
````

### Alternative approach — convert to list, sort, convert back to tuple

````python
t1 = (10, -4, 12, 26, 16, -6, 0, 15)
print(t1, type(t1))

l1 = list(t1)
l1.sort()               # sort the list in place
ex = tuple(l1)          # convert back into tuple
print(ex, type(ex))
````

## Inner / Nested Tuple

The process of defining one tuple inside another tuple is called an **Inner** or **Nested** tuple.

**Syntax:**
```
tup1 = (val1, val2, (val21, val22, ... val2n))
```

- `(val1, val2, ...)` is called the **outer** tuple.
- `(val21, val22, ... val2n)` is called the **inner** or **nested** tuple.

**Rules:**
- All the pre-defined functions of tuple can be applied on the inner/nested tuple as well.
- On the inner (or nested) tuple, we can perform indexing or slicing operations.


````python
sf = (10, "RS", (17, 18, 16), (78, 68, 79), "onetwo")
print(sf, type(sf))
print(sf[0])
print(sf[2], type(sf[2]), type(sf))
print(sf[0:3])
````

### Combination of List and Tuple
When a **list** is nested inside a tuple, the inner list is still **mutable**, even though the outer tuple is not.


````python
sf = (10, "RS", [17, 18, 16], (78, 68, 79), "onetwo")
print(sf, type(sf))
print(sf[2], type(sf[2]))

sf[2].append(12)     # allowed - the inner object is a list (mutable)
print(sf[2])
print(sf)

# sf[3].append(12)   -> AttributeError: 'tuple' object has no attribute 'append'
````

### Modifying nested mutable elements inside a tuple

````python
t1 = (10, "Ross", [16, 18, 17], ('AI', 'CSE', 'DS'))

t1[2][1] = 18          # allowed, since t1[2] is a list
print(t1)

k = sorted(t1[3])      # sorted() on the inner tuple returns a list
print(k)

# t1[3] = 808          -> TypeError: 'tuple' object does not support item assignment
````

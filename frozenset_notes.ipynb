# Frozenset


**Note:** The functionality of `frozenset` is exactly similar to `set`, but an object of `set` belongs to **both** mutable and immutable, whereas an object of `frozenset` belongs **only** to immutable.


**Syntax**

```python
frozensetobj = frozenset(object)
```

- `frozenset` is one of the pre-defined classes and is treated as a set data type.
- The `frozenset` data type is used to store multiple values, either of similar type or different type or both, in a single object with **unique** values.
- The elements of a `frozenset` must be obtained from a different object such as a `set`, `tuple`, `list`, etc.


### Key characteristics

- An object of `frozenset` never maintains insertion order, because we can't determine in advance which element of the frozenset object will be shown first.
- On a `frozenset` we can't perform indexing and slicing operations, because a frozenset object never maintains insertion order.
- An object of `frozenset` belongs to immutable, because a frozenset object does not support item reassignment (add / remove / modify elements).
- We can create two types of frozenset objects:
    1. Empty frozenset
    2. Non-Empty frozenset

**1. Empty frozenset** &rarr; An empty frozenset is one which does not contain any elements, where length is `0`.

```python
frozensetobj = frozenset()
```

**2. Non-Empty frozenset** &rarr; A non-empty frozenset contains elements, where length is `> 0`.

```python
frozensetobj = frozenset(object)
```


## Example 1

````python
s1 = {30, 40, 50}
print(type(s1))

fs1 = frozenset(s1)
print(fs1, type(fs1))
````

## Example 2 — empty frozenset

````python
fs2 = frozenset()
print(fs2, type(fs2))
print(len(fs2))
````

## Example 3 — frozenset does not support indexing, slicing, item assignment or item deletion

````python
s1 = {30, 40, 50}
fs3 = frozenset(s1)
fs3[0]
````

````python
fs3[0:2]
````

````python
fs3[0] = 23   # item reassignment
````

````python
del fs3[0]   # item deletion
````

````python
del fs3[0:1]   # slice deletion
````

Deleting the **entire** frozenset object (not an individual item) is possible, since that operation is on the reference/name, not on the frozenset's content.

````python
del fs3   # possible, deletes the whole object
print(fs3)
````

# Pre-defined functions in frozenset

`frozenset` contains the following functions:

1. `copy()`
2. `isdisjoint()`
3. `issuperset()`
4. `issubset()`
5. `union()`
6. `intersection()`
7. `difference()`
8. `symmetric_difference()`


### `copy()`

````python
fs1 = frozenset({10, 20, 30})
print(fs1, type(fs1), id(fs1))
````

````python
fs2 = fs1.copy()   # deep copy
print(fs2, type(fs2), id(fs2))
````

**Note:** In general, the content of an immutable object is not possible to copy into a truly separate object of the same type (in the case of any datatype). With `frozenset`, we are able to "copy" its content to another frozenset object — but here the original frozenset object and the copied frozenset object contain the **same memory address** (`id`), and it is **not at all possible** to modify or change either object's content, since both are immutable.

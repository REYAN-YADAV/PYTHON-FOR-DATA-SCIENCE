### 4. Interpreted

When we develop any Python program, we must give some file name with an extension `.py` (`filename.py`).

When we execute Python program, two processes take place internally:

* **a) Compilation process**
* **b) Execution process**

In **Compilation Process**, the Python source code submitted to Python compiler and it reads the source code (line by line), checks for error by verifying syntaxes, and if no errors found, then Python compiler converts into intermediate code called **Byte Code** with an extension `.pyc`. If errors are found in source code, then we get error display on the console.

In **Execution Process**, the **PVM** reads the Python intermediate code (`Byte code`) line by line and converts it into machine code (executable or binary code) and it is read by OS and processor and finally gives result.

Hence, in Python programming execution, compilation process and execution process taking place line by line conversion and it is one of the **Interpretation Based** programming language.

> **Definition of PVM (Python Virtual Machine):**
> PVM is one program in Python software whose role is to read line-by-line byte code and convert it into machine understandable code (executable or binary code).

$$\text{Source Code} \xrightarrow{\text{Compiler Process}} \text{Byte Code (.pyc)} \xrightarrow[\text{PVM}]{\text{Execution Process}} \text{Machine Code}$$

---

### **Q. What is Byte Code?**

* Compiled code of Python source code is called byte code.
* It is intermediate code of Python execution environment.
* Its extension is `.pyc`.
* It is platform independent.
* It is understandable by PVM.
* Byte code is collection of mnemonics.

### **Difference between Interpreter and Compiler**

| Interpreter | Compiler |
| --- | --- |
| Executes program by taking one statement at a time. | Translates the entire program at once into machine code. |
| Error reported as soon as first error is encountered; won't show the next set of errors if existing one isn't solved. | Errors are reported after entire program is checked. |
| **Easier** to debug. | **Difficult** to debug. |
| No intermediate object code is generated, hence memory efficient. | Generates intermediate object code which further requires linking, hence requires more memory. |
| Overall execution is **slower**. | Overall execution is **faster**. |

## **Importance of Identifiers or Variable**

* To process values which are present in memory spaces, programmers must create memory space with **distinct names**. These distinct names make us to identify the values present in memory space and hence they are called **Identifiers**.
* The identifier values are changing / varying during program execution and hence identifiers are called **Variables**.

### **Rules for Using Identifiers or Variables in Python Programs**

To use variables in Python programs, we follow the following rules:

* **Rule 1:** The variable names are a combination of Alphabets, Digits, and Special Symbol (`_` underscore only).
* **Rule 2:** The first letter of variable name **must start with either Alphabet or Underscore (`_`) only**.

*Examples:*

```python
sal = 56        # Valid
_sal = 56       # Valid
_Sal = 57       # Valid
sal9 = 57       # Valid
__sal = 57      # Valid
123 = 58        # Invalid (SyntaxError)

```
* **Rule 3:** Within the variables, no special symbols are allowed except underscore (`_`).
* **Rule 4:** No keywords to be used as variable name (because keywords are reserved words and they give special meaning to the compilers).
```python
while = 56    # Invalid ✗
While = 56    # Valid ✓
for = 56      # Invalid ✗
For = 56      # Valid ✓
False = 56    # Invalid ✗
false = 56    # Valid ✓

```


* **Rule 5:** All the variable names are case sensitive.
```python
age = 98
Age = 99
aGe = 99
AGE = 100

print(age, Age, aGe, AGE)

```

### **Syntax 5:**

In Python, we have 3 types of format specifiers. They are:

* **a)** `%d` $\rightarrow$ for `int`

* **c)** `%s` $\rightarrow$ for `str`

```python
sno = 10
sname = "Raj"
marks = 24.5

print("Sno is %d and name is %s and marks is %0.2f" % (sno, sname, marks))

lst = [10, "Branch", 34.5, True]
print("Content of list = {} ".format(lst))
print("Content of list = %s" % str(lst))

```

### List Category Data Types (Collection Data Types / Data Structures)

The purpose of List Category Data Types is: *"To store multiple values, either of the same type, or different type, or both types, in a single object, with unique and duplicate values allowed."*

We have two data types in the List Category:
1. **list** — mutable

## 1. List
### Properties of `list`
`list` is one of the pre-defined classes and is treated as a list data type.

The purpose of the list data type is: *"To store multiple values of the same type, different types, or both types in a single object, with unique and duplicate values allowed."*

To store values/elements in a list, they must be enclosed (written) inside square brackets `[ ]`, and values must be separated by commas.

**Syntax:**
```python
varname = [val1, val2, ..., valn]
```

````python
list1 = [10, 20, 15, 45, 10, 12, 999]
print(list1, type(list1))
````

````python
list2 = [10, "Ryan", 44.4, True, 2 + 3j]
print(list2)
````

An object of `list` maintains **insertion order**.

On the object of `list`, we can perform both **Indexing** and **Slicing** operations.

````python
list2 = [10, "Ray", 44.4, True, 2 + 3j]
print(list2[0])
print(list2[0:3])
````

An object of `list` belongs to **mutable**.

````python
list2 = [10, "Rey", 44.4, True, 2 + 3j]
print(list2, type(list2), id(list2))
````

### Index Based Update

````python
list2[0] = 100   # index based update
print(list2, id(list2))
````

### Slice Based Update

````python
list2[1:3] = [300, 55.5]   # slice based update
print(list2, id(list2))
````

We can create two types of list objects. They are:
1. **Empty list**
2. **Non-Empty list**

**Empty list** → An empty list is one which does not contain any elements, and whose length is `0`.
- Syntax 1 → `varname = []`
- Syntax 2 → `varname = list()`

**Non-Empty list** → A non-empty list is one which contains elements, and whose length is `!= 0`.
- Syntax 1 → `varname = [val1, val2, ...]`
- Syntax 2 → `varname = list(object)`

````python
S = "python"
print(S, type(S))
list1 = list(S)
print(list1, type(list1))
````

`str` → `list` conversion stores the sequence of characters in a `str` object into a single list object.

````python
lst1 = [10, 20, 30, 40, 255]
b = bytes(lst1)
print(b, type(b))
lst1 = list(b)
print(lst1, type(lst1))
````

### 1. `append()`
**Syntax:** `listobj.append(value)`

This function is used for adding an element (value) at the end of the list object — called append.

````python
list1 = [10, "Rey", 22]
print(list1, type(list1), id(list1))
list1.append("python")
list1.append(2 + 2j)
print(list1, id(list1))
````

````python
list1 = []
list1.append(100)
list1.append("Rama")
list1.append("Krishna")
print(list1)
````

````python
try:
    list1 = [22, "Reym", 34]
    list1.append(22, "Reyw", 34)   # more than 1 argument
except TypeError as e:
    print("TypeError:", e)
````

### 2. `insert()`
**Syntax:** `listobj.insert(index, value)`

This function is used for adding (inserting) the value at a specified index.

````python
list1 = [10, "Rossom", 22.2]
list1.insert(1, "python")
print(list1)
````

Here `index` can be either positive or negative.

````python
list1 = [10, "Ross", 22]
list1.insert(-1, "python")
print(list1)
list1.insert(-3, "vivek")
print(list1)
````

When we specify an invalid `+ve` index, then the value is inserted at the last (end).
When we specify an invalid `-ve` index, then the value is inserted at the first (begin).

````python
list1 = [10, "Ross", 22]
list1.insert(200, "python")
print(list1)
list1.insert(-200, "HTML")
print(list1)
````

### 3. `remove()` — Based on value
**Syntax:** `listobj.remove(value)`

This function is used for removing the first occurrence of the specified value. If the specified value does not exist in the list object, then we get a `ValueError`.

````python
list1 = [10, "Ross", 22]
list1.remove("Ross")
print(list1)
list1.remove(10)
print(list1)
list1.remove(22)
print(list1)
````

````python
try:
    list1.remove("python")
except ValueError as e:
    print("ValueError:", e)
````

````python
list1 = [10, 20, 30, 10, 40]
list1.remove(10)
print(list1)
list1.remove(10)
print(list1)
````

````python
try:
    [].remove(10)
except ValueError as e:
    print("ValueError:", e)
````

### 4. `pop(index)`
**Syntax:** `listobj.pop(index)`

This function is used for removing the value from a list object by passing the index. Here index can be either `+ve` or `-ve`. When we specify an invalid index, then we get `IndexError`.

````python
list1 = [10, 20, 30, 10, 20, 40, 50, 45]
list1.pop(3)
print(list1)
list1.pop(-4)
print(list1)
````

````python
try:
    list1.pop(120)
except IndexError as e:
    print("IndexError:", e)

try:
    [].pop(0)
except IndexError as e:
    print("IndexError:", e)

try:
    [].pop(-1)
except IndexError as e:
    print("IndexError:", e)
````

### 5. `pop()`
**Syntax:** `listobj.pop()`

This function is used for removing the last element of a non-empty list object. If we use this function on an empty list object, then we get `IndexError`.

````python
lst = [10, "Ross", 34, True, 3 + 2j]
print(lst.pop())
print(lst)
print(lst.pop())
print(lst)
print(lst.pop())
print(lst)
````

````python
[10, 20, 30].pop()
````

````python
print(list("PYTHON").pop())
print(list("python").pop(-2))
print(list("python").pop(-len("python")))
````

````python
try:
    list("python").pop(len("python"))
except IndexError as e:
    print("IndexError:", e)
````

### 6. `clear()`
**Syntax:** `listobj.clear()`

This function is used for removing all the elements from a non-empty list object. When we use this function on an empty list object, then we get no space error / `None`.

````python
lst = [10, "Rossum", 34.5, True]
lst.clear()
print(lst)
print(len(lst))
````

### `del` operator
- **Syntax 1:** `del object[index]`
- **Syntax 2:** `del object[start : stop : step]`
- **Syntax 3:** `del object`

Syntax 1 → Removes the element of any mutable object based on indexing.
Syntax 2 → Removes the element of any mutable object based on slicing.
Syntax 3 → Removes the entire object of type mutable or immutable.

````python
lst = [10, "Ross", 34.5, 2 + 3j]
del lst[3]   # based on index
print(lst)
del lst[1]
print(lst)
````

````python
lst = [10, "Ross", 34.5, 2 + 3j]
del lst[0:3]   # based on slicing
print(lst)
````

````python
lst = [10, "Ross", 34.5, 2 + 3j]
del lst   # complete object removed
try:
    print(lst)
except NameError as e:
    print("NameError:", e)
````

````python
S = "python"
print(S, type(S))
try:
    del S[0]
except TypeError as e:
    print("TypeError:", e, "(immutable)")

try:
    del S[0:4]
except TypeError as e:
    print("TypeError:", e, "(immutable)")
````

````python
S = "python"
del S   # remove complete object
try:
    print(S)
except NameError as e:
    print("NameError:", e)
````

### 7. `copy()`
In Python programming, we have two types of copy techniques. They are:
1. Shallow Copy
2. Deep Copy

#### 1. Shallow Copy
To implement shallow copy, we use `copy()`.

**Syntax:** `listobj2 = listobj1.copy()`

The properties of Shallow Copy are:
1. Initial content of both the objects are same.
2. The memory address of both the objects are different.
3. The modification on both the objects are independent (whatever changes we do on one object, those changes will not reflect on the other object).

````python
list1 = [10, "Ross"]
print(list1, id(list1))
list2 = list1.copy()   # shallow copy
print(list2, id(list2))
````

````python
list1.append("python")
print(list1)
list2.insert(1, "Guddu")
print(list2)
print(list1, id(list1))
print(list2, id(list2))
````

#### 2. Deep Copy
The properties of Deep Copy are:
1. Initial content of both objects are same.
2. The memory address of both objects are same.
3. The modification on both the objects are dependent (whatever changes we do on the object, those changes will reflect on the other object).

To implement deep copy, we use the assignment operator (`=`).

**Syntax:** `listobj2 = listobj1`

````python
lst1 = [10, "Ross"]
print(lst1, id(lst1))
lst2 = lst1   # deep copy (same object)
print(lst2, id(lst2))
````

````python
lst1.append("python")
print(lst1, id(lst1))
print(lst2, id(lst2))
lst2.insert(1, "Guddu")
print(lst2, id(lst2))
````

### `enumerate()`
`enumerate()` gives an `(index, value)` pair for every element in an iterable — useful for going through a list (or `str`) along with its index.

````python
S = "mississippi"
for ind, char in enumerate(S):
    print(ind, "-->", char)
````

## 8. `sort()` Method
`sort()` arranges the elements of a list **in place** (it modifies the original list, and returns `None` — so `print(lst.sort())` prints `None`).

**Syntax 1 (ascending — default):**
```python
listobj.sort()
```
**Syntax 2 (ascending, explicit):**
```python
listobj.sort(reverse=True)   # descending
listobj.sort(reverse=False)  # ascending (default)
```

````python
lst = [10, 2, 31, -4, 0, 45, 34, 22, 11]
print(lst, id(lst))
````

````python
lst.sort()
print(lst, id(lst))   # same id -> sorted in place, ascending
````

````python
lst.reverse()
print(lst, id(lst))   # reversed in place, same id
````

### Example — `sort(reverse=True/False)`

````python
lst2 = [10, 2, 31, -4, 0, 45, 34, 22, 11]
print(lst2, id(lst2))
````

````python
lst2.sort(reverse=False)   # ascending
print(lst2, id(lst2))
````

````python
lst2.sort(reverse=True)    # descending
print(lst2, id(lst2))
````

### Example — sorting a list of strings

````python
lst3 = ["Biden", "Adams", "Trump", "Zoom", "Trav"]
lst3.sort(reverse=True)
print(lst3)
````

### `sort()` fails on mixed, non-comparable types
Python cannot compare `str` and `int` with `<`, so sorting a mixed-type list raises a `TypeError`.

````python
lst4 = [10, "gods", 4.57, True]
print(lst4, id(lst4))
try:
    lst4.sort()
except TypeError as e:
    print("TypeError:", e)
````

## 9. `extend()` Method
`extend()` adds **all the elements** of one list to another (merges them one at a time), unlike `append()` which would add the *whole list* as a single nested element.

**Syntax 1:**
```python
listobj1.extend(listobj2)
```
**Syntax 2 (equivalent, using `+`):**
```python
list1 = list1 + list2 + list3 ...
```

````python
lst1 = [10, 20, 30]
lst2 = ['RS', 'TR', 'DR', 'py']
lst1.extend(lst2)
print(lst1)
````

### Reversed order — `extend` called on `lst2` instead

````python
lst1 = [10, 20, 30]
lst2 = ['RS', 'TR', 'DR', 'py']
lst2.extend(lst1)
print(lst2)
````

### `extend()` takes exactly ONE argument
Passing more than one list raises a `TypeError`. Use `+` (Syntax 2) instead to merge multiple lists at once.

````python
lst1 = [10, 20, 30]
lst2 = ['RS', 'TR', 'DR']
lst3 = [1.2, 2.3, 4.5]
try:
    lst1.extend(lst2, lst3)
except TypeError as e:
    print("TypeError:", e)
````

````python
# Using the + operator (Syntax 2) to merge multiple lists at once
lst1 = [10, 20, 30]
lst2 = ['RS', 'TR', 'DR']
lst3 = [1.2, 2.3, 4.5]
lst1 = lst1 + lst2 + lst3
print(lst1)
````

### `extend()` with a string
A string is iterable, so `extend()` adds it **character by character**, not as a single element.

````python
lst = [10, 20, 30]
lst.extend("python")
print(lst)
````

## 10. Slice-based Copy vs `=` Assignment
`=` assignment does **not** create a new list — both names point to the *same* object (same `id`). Slicing (`lst[:]`) creates a genuinely new list object (a shallow copy) — different `id`.

````python
lst1 = [10, 20, 30, 40]
lst2 = lst1          # NOT a copy — same object
print(lst1, id(lst1))
print(lst2, id(lst2))
````

````python
lst1 = [10, 20, 30, 40]
lst2 = lst1[::2]     # slice -> new object
print(lst2)
````

````python
lst = [10, 20, 30]
lst2 = lst[:]         # full slice -> shallow copy, new id
print(lst, id(lst))
print(lst2, id(lst2))
````

## 11. Inner / Nested Lists
> The process of defining one list inside another list is called an **inner (nested) list**.

**Syntax:**
```python
listobj = [val1, val2, ..., [val11, val12, ...valn1],
                             [val21, val22, ...valn2], ... valn]
```
Here `[val11, val12, ...valn1]` and `[val21, val22, ...valn2]` are the **inner lists**, and the whole `listobj` is the **outer list**.

````python
lst = [10, "srinivas", [18, 16, 19], [70, 77, 80], "sweet"]
for val in lst:
    print(val, type(val), type(lst))
````

Indices, counted from both ends, for `lst = [10, "srinivas", [18,16,19], [70,77,80], "sweet"]`:

| index |  0  |  1  |  2  |  3  |  4  |
|---|---|---|---|---|---|
| value | 10 | 'srinivas' | [18,16,19] | [70,77,80] | 'sweet' |
| neg index | -5 | -4 | -3 | -2 | -1 |

## 12. Indexing & Slicing Nested Lists

````python
lst = [10, "Reyan", [18, 16, 19], [70, 77, 80], "sweet"]
print(lst[1])       # 'Reyan'
print(lst[2])       # [18, 16, 19]
print(lst[-2])      # [70, 77, 80]  -> same as lst[3]
print(lst[-1])      # 'sweet'
````

````python
# Indexing INTO an inner list: lst[outer_index][inner_index]
print(lst[2][0])    # first element of the 3rd item -> 18
print(lst[2][-1])   # last element of that inner list -> 19
print(lst[-2][1])   # 2nd element of the 2nd-last item -> 77
print(lst[-2][-1])  # last element of that inner list -> 80
````

````python
# Slicing an inner list: lst[outer_index][start:stop]
print(lst[3][0:2])   # [70, 77]
print(lst[3][::2])   # [70, 80]
print(lst[-2][0:2])  # [70, 77]  (lst[-2] is the same as lst[3])
print(lst[-2][::2])  # [70, 80]
````

> **On inner list objects we can perform both indexing and slicing operations, and every list method (`append`, `insert`, `remove`, `sort`, `reverse`, `del`, etc.) works on them too.**

## 13. Modifying Inner Lists

````python
lst = [10, "Reyan", [18, 16, 19], [70, 77, 80], "sweet"]
lst[2][1] = 20               # modify one element of an inner list
print(lst)
````

````python
lst[2].append(15)            # append to an inner list
print(lst)
````

````python
lst[3].insert(1, 80)         # insert 80 at index 1 of the inner list
print(lst)
````

````python
lst[-3].sort()                # sort inner list in place (ascending)
print(lst)
````

````python
lst[-2].sort(reverse=True)    # sort inner list descending
print(lst)
````

````python
del lst[2][1:]                # delete a slice from an inner list
print(lst)
````

````python
lst[-2].remove(80)            # remove first matching value from inner list
print(lst)
````

````python
del lst[2]                    # delete the whole inner list (element) from outer list
print(lst)
````

````python
lst.insert(2, [15, 18, 13])   # insert a brand-new inner list back in
print(lst)
````

### Out-of-range assignment on an inner list raises `IndexError`

````python
lst = [10, "Reyan", [18, 16, 19], [70, 77, 80], "sweet"]
try:
    lst[2][200] = 20
except IndexError as e:
    print("IndexError:", e)
````

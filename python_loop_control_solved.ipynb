## Transfer Flow Statements

The purpose of transfer flow statements is to transfer control of the PVM (Python Virtual Machine) from one part of a Python program to another part of the program.

There are 4 types of transfer flow statements:
1. `break`
2. `continue`
3. `pass`
4. `return`

## 1. `break`

- `break` is a **keyword**.
- The purpose of the `break` statement is to terminate the execution of a loop logically when a certain condition is satisfied, and program control comes out of the corresponding loop and executes the other statements in the program.
- When `break` takes place inside a `for` loop or `while` loop, the corresponding `else` block does **not** execute (because the loop did not become false on its own) — but the rest of the program's statements still execute.
- `break` must be used inside a loop, otherwise you get a `SyntaxError`.

**Syntax 1**
```python
for varname in iterable_object:
    if (test_cond):
        break
```

**Syntax 2**
```python
while (test_cond_1):
    if (test_cond_2):
        break
```

## 2. `continue`

- `continue` is a **keyword**.
- The `continue` statement is used to make the PVM jump to the top of the loop, **without** executing the statements that come after `continue` for that current iteration only.
- `continue` must always be used inside a loop.
- When `continue` runs inside a loop, the `else` part of the corresponding loop still executes, provided the loop condition eventually becomes `False`.

**Syntax**
```python
for varname in iterable_object:
    if (test_cond):
        continue
    statement-1
```

```python
while (test_cond_1):
    if (test_cond_2):
        continue
    statement-1
    statement-2
```

### Program demonstrating `continue`

Requirement: display `'pyhn'` from `s = "python"` (skip `'t'` and `'o'`).

````python
s = "python"
print("For loop version:")
for ch in s:
    if ch == 't' or ch == 'o':
        continue
    print(ch, end="")
else:
    print("\nI am from else part of for loop")

print()
print("While loop version:")
i = 0
while i < len(s):
    if s[i] == 't' or s[i] == 'o':
        i = i + 1
        continue
    print(s[i], end="")
    i = i + 1
else:
    print("\nI am from else part of while loop")
````

### String Data Order

Given a line of text, separate letters and digits into two lists, then: join the letters back into a string (sorted), and split the digits into even/odd lists sorted ascending / descending.

````python
line = "Python420and71"   # normally: input("Enter line of Text: ")

strlist = list()
numlist = list()
for ch in line:
    if ch.isalpha():
        strlist.append(ch)
    elif ch.isdigit():
        numlist.append(int(ch))
    else:
        pass

print("String characters:", strlist)
print("Number characters:", numlist)

strline = ''
strline = strline.join(strlist)
print("Joined string characters:", strline)

ed = []
od = []
for d in numlist:
    if d % 2 == 0:
        ed.append(d)
    else:
        od.append(d)

print("Even list = {}".format(ed))
print("Odd list  = {}".format(od))

ed.sort()
print("Sorted even list (ascending)  = {}".format(ed))
ed.sort(reverse=True)
print("Sorted even list (descending) = {}".format(ed))

od.sort()
print("Sorted odd list (ascending)   = {}".format(od))

strdata = ''
strdata = strdata.join(sorted(strlist))
print("Sorted string characters = {}".format(strdata))
````

## Nested (Inner) Loops

The process of defining one loop inside of another loop is called an inner or nested loop.

The execution process of a nested loop is: for every value of the outer loop, the inner loop executes/repeats for the full number of times, until the inner loop condition becomes false.

We can nest loops in 4 ways:
1. `for` loop inside `for` loop
2. `while` loop inside `while` loop
3. `for` loop inside `while` loop
4. `while` loop inside `for` loop

### Generate the list of prime numbers within `n`, where `n` is a +ve integer

````python
def primes_upto(n):
    primes = []
    for num in range(2, n + 1):
        is_p = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_p = False
                break
        if is_p:
            primes.append(num)
    return primes

n = 10
print("Prime numbers up to {}: {}".format(n, primes_upto(n)))
````

### Accept a list of values and display only the primes

````python
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True

values = [2, 4, 7, 9, 11, 15, 17, 20, 23]
primes_only = [v for v in values if is_prime(v)]
print("Given values:", values)
print("Primes only :", primes_only)
````

## Name Validation

Write a program which validates the name of a person, place, or thing (each word must contain only alphabetic characters).

````python
def validate_name(name):
    words = name.split()
    res = True
    for word in words:
        if not word.isalpha():
            res = False
            break
    return res

for candidate in ["John Smith", "Mumbai123", "New York", "Product#1"]:
    if validate_name(candidate):
        print("{} is a valid name".format(candidate))
    else:
        print("{} is not a valid name".format(candidate))
````

## Student Marks Report System

Accept student number, student name, and marks in three subjects — C-lang (100), C++ (100), Python (100).

- **Total marks** = C + C++ + Python; **Percentage** = (Total / 300) × 100
- **Fail** condition: student scored less than 40 in *any* of the three subjects.
- **Pass** — division decided by total marks:
  - **Distinction**: total in 250–300 (inclusive)
  - **First division**: total in 200–249
  - **Second division**: total in 150–199
  - **Third division**: total in 120–149

### Input validation helpers (student number, name, marks)

Each of these loops normally reads from `input()`; here a list of simulated user entries is fed in so the validation logic can be demonstrated end-to-end.

````python
def get_valid_student_number(sample_inputs):
    idx = 0
    while True:
        sno = sample_inputs[idx]; idx += 1     # normally: input("Enter Stud Num: ")
        if sno.isdigit():
            if int(sno) > 0:
                return int(sno)
            else:
                print("{} Invalid input for Student Number - try again".format(sno))
        else:
            print("{} - Student Number must not be a non-integer, try again".format(sno))

def get_valid_name(sample_inputs):
    idx = 0
    while True:
        sname = sample_inputs[idx]; idx += 1   # normally: input("Enter student name: ")
        if sname.replace(" ", "").isalpha():
            return sname
        else:
            print("{} is invalid for student name, try again".format(sname))

def get_valid_marks(subject, sample_inputs):
    idx = 0
    while True:
        raw = sample_inputs[idx]; idx += 1     # normally: input("Enter {} Marks: ".format(subject))
        if raw.isdigit() and 0 <= int(raw) <= 100:
            return int(raw)
        else:
            print("{} is invalid input for {} marks".format(raw, subject))

print("Validated student number:", get_valid_student_number(["-5", "abc", "101"]))
print("Validated student name  :", get_valid_name(["John123", "John Smith"]))
print("Validated C marks       :", get_valid_marks("C-lang", ["150", "-3", "88"]))
````

### Full report generator

````python
def get_student_record(sno, sname, cm, cppm, pypm):
    total = cm + cppm + pypm
    percentage = (total / 300) * 100

    if cm < 40 or cppm < 40 or pypm < 40:
        result = "Fail"
        division = "-"
    else:
        result = "Pass"
        if 250 <= total <= 300:
            division = "Distinction"
        elif 200 <= total <= 249:
            division = "First"
        elif 150 <= total <= 199:
            division = "Second"
        elif 120 <= total <= 149:
            division = "Third"
        else:
            division = "-"
            result = "Fail"

    return {
        "Student No": sno, "Name": sname,
        "C-lang": cm, "C++": cppm, "Python": pypm,
        "Total": total, "Percentage": round(percentage, 2),
        "Result": result, "Division": division,
    }

students = [
    (1, "Ravi Kumar",   85, 90, 95),
    (2, "Anita Sharma", 35, 60, 70),
    (3, "Manoj Verma",  60, 55, 50),
    (4, "Priya Singh",  78, 72, 80),
]

for sno, sname, cm, cppm, pypm in students:
    record = get_student_record(sno, sname, cm, cppm, pypm)
    for k, v in record.items():
        print("{}: {}".format(k, v))
    print("-" * 40)
````

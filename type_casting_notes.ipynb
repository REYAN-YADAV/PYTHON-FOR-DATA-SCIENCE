# Type Casting Techniques in Python


The process of converting one possible type of value into another type of value is called **type casting**.

In Python programming, we have **5 fundamental type casting techniques**. They are:

1. `int()`
2. `float()`
3. `bool()`
4. `complex()`
5. `str()`


## 1. `int()`

`int()` is used for converting one possible type of value into `int` type value.

**Syntax:** `varname = int(float | bool | complex | str)`


### Ex 1. float to int → possible

```python
a = 10.23
print(a, type(a))
```

```python
b = int(a)
print(b, type(b))
```

```python
print(int(99.99))
print(int(0.99))
```

### Ex 2. bool to int → possible

```python
a = True
print(a, type(a))
b = int(a)
print(b, type(b))
```

```python
a = False
b = int(a)
print(b, type(b))
```

### Ex 3. complex to int → **NOT possible**

```python
a = 2 + 3j
print(a, type(a))
b = int(a)  # raises TypeError
```

### Ex 4. str to int

Depends on what the string contains:

**Case 1: str "int" into int → possible**

```python
a = "12"
print(a, type(a))
b = int(a)
print(b, type(b))
```

**Case 2: str "float" into int → NOT possible**

```python
a = "12.34"
print(a, type(a))
b = int(a)  # raises ValueError
```

**Case 3: str "complex" into int → NOT possible**

```python
a = "2+3j"
b = int(a)  # raises ValueError
```

**Case 4: str "bool" into int → NOT possible**

```python
a = "True"
b = int(a)  # raises ValueError
```

**Case 5: pure str into int → NOT possible**

```python
a = "8"
b = int(a)
print(b)

a = "$"
b = int(a)  # raises ValueError
```

## 2. `float()`

`float()` is used for converting one possible type value into `float` type value.

**Syntax:** `varname = float(int | bool | str | complex)`


### Ex 1. int type into float type → possible

```python
a = 12
b = float(a)
print(b, type(b))
```

### Ex 2. bool type to float type → possible

```python
a = True
b = float(a)
print(b, type(b))
```

### Ex 3. complex to float type → **NOT possible**

```python
a = 3 + 3.5j
b = float(a)  # raises TypeError
```

### Ex 4. str type to float type

**Case 1: str int into float → possible**

```python
a = "123"
b = float(a)
print(b)
```

**Case 2: str float into float → possible**

```python
a = "12.34"
b = float(a)
print(b)
```

**Case 3: str complex into float → NOT possible**

```python
a = "2+3j"
b = float(a)  # raises ValueError
```

**Case 4: str bool into float → NOT possible**

```python
a = "True"
b = float(a)  # raises ValueError
```

**Case 5: pure str into float → NOT possible**

```python
a = "python"
b = float(a)  # raises ValueError
```

## 3. `bool()`

`bool()` is used for converting one possible type value into `bool` type value.

**Syntax:** `varname = bool(int | float | complex | str)`

- All **non-zero** values are considered as `True`
- All **zero** values are considered as `False`

Therefore, every type — `str`, `int`, `float` and `complex` — can be converted into the `bool` data type.

### Examples

```python
a = 123
b = bool(a)
print(b)
```

```python
a = 0
b = bool(a)
print(b)
```

```python
a = 12.45
b = bool(a)
print(b)
```

```python
a = 0.0
b = bool(a)
print(b)
```

```python
a = 3 + 2j
b = bool(a)
print(b)
```

```python
a = 0 + 0j
b = bool(a)
print(b)
```

```python
a = "True"
b = bool(a)
print(b)
```

```python
a = "python"
b = bool(a)
print(b)
```

### Misc questions

```python
print(bool(int(float("123"))))
```

```python
print(bool(0x0))       # hex 0
```

```python
print(bool(0o40 - 0b1))  # octal 40 minus binary 1
```

```python
print(bool(1/2))
```

## 4. `complex()`

`complex()` is used for converting one possible type of value into `complex` type value.

**Syntax:** `varname = complex(int | float | str | bool)`


### Ex 1. int type to complex type → possible

```python
a = 10
b = complex(a)
print(b)
```

### Ex 2. float type to complex type → possible

```python
a = 10.2
b = complex(a)
print(b)
```

### Ex 3. bool type to complex type → possible

```python
a = True
b = complex(a)
print(b)

a = False
b = complex(a)
print(b)
```

### Ex 4. str type to complex type

**Case 1: str int into complex → possible**

```python
a = "10"
b = complex(a)
print(b)
```

**Case 2: str float into complex → possible**

```python
a = "10.5"
b = complex(a)
print(b)
```

**Case 3: str complex into complex → possible (no spaces allowed)**

```python
a = "1+2j"
b = complex(a)
print(b)
```

```python
a = "1 + 2j"   # spaces make it malformed
b = complex(a)  # raises ValueError
```

**Case 4: str bool into complex → NOT possible**

```python
a = "True"
b = complex(a)  # raises ValueError
```

**Case 5: pure str into complex → NOT possible**

```python
a = "python"
b = complex(a)  # raises ValueError
```

## 5. `str()`

`str()` is used for converting **all** types of value into `str` type value.

**Syntax:** `varname = str(int | float | bool | complex)`


### Examples

```python
a = 12
b = str(a)
print(b, type(b))
```

```python
a = 12.24
b = str(a)
print(b, type(b))
```

```python
a = True
b = str(a)
print(b, type(b))
```

## Mutable and Immutable Objects

### Mutable object
A **mutable object** is one which allows us to modify the values at the **same address**.

**Examples:** `list`, `set`, `dict`, `bytearray`

### Immutable object
An **immutable object** is one which satisfies the following properties:

- An immutable object **never allows** us to modify at the same address (the modified value is placed at a **new address**).
- An immutable object **never allows** us to perform item assignment.

**Examples:** `int`, `float`, `bool`, `complex`, `str`, `bytes`, `range`, `tuple`, `set` *(frozenset)*, `NoneType`
